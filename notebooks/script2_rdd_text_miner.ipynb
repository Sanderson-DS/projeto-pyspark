# Segundo Script RDD - Text Miner

```python
# Versão do Spark Context
from pyspark.sql import SparkSession

spark = SparkSession.builder \
    .appName("TextMinerRDD") \
    .getOrCreate()

print(spark.version)
```

```python
# Carregando os dados sobre economia
data = spark.sparkContext.textFile("../data/regiao-sul.txt,../data/regiao-norte.txt")
```

```python
data.take(10)
```

```python
# separação das palavras com o uso da função lambda e spli
palavras = data.flatMap(lambda x: x.split(" ")) 
```

```python
# exibição da coleção de palavras
for plv in palavras.collect():
    print(plv)
```

```python
# Função que transforma em minusculas todas as palavras

def func1(x):
    minuscula = x.lower()
    return minuscula

allpalminusculas = palavras.map(lambda x: func1(x))
```

```python
# Exibe a colecao de dados
allpalminusculas.collect()
```

```python
# Cria a lista de stopwords que será filtrada
stopwords =['e','do','das','na','há','ser','são','para','por', 'as','em','no','nas','uma','que','como','com','','os','o','a','da','de','nas','é', 'ao', 'ao', 'seu', 'seus', 'sua', 'suas', 'meu', 'meus', 'minha', 'minhas', 'esses', 'essas', 'este', 'estes', 'esta', 'estas', 'aquele', 'aqueles', 'aquela', 'aquelas',',','.']
filtro_palavras = allpalminusculas.filter(lambda x: x not in stopwords)
```

```python
# exibe a coleção de palavras filtradas
filtro_palavras.collect()
```

```python
# Preparando o contador para somar a quantidade de palavras, colocando o número 1 ao lado das palavras
freqpal = filtro_palavras.map(lambda x: (x, 1))
```

```python
# Frequencia de palavras
freqpal.collect()
```

```python
# utilizando a função reducekey para somar o contador
from operator import add
totalpal = freqpal.reduceByKey(add)
```

```python
# mostrar o plano de execução
totalpal.toDebugString()
```

```python
# exibindo a contagem ao lado da palavra
totalpal.collect()
```

```python
# Realizando um for interativo para colocar as palavras em ordem decrescente
for i in totalpal.sortBy(keyfunc= lambda x:x[1], ascending=False).collect():print(i)
```

```python
# fechar o spark 
spark.stop()
```
