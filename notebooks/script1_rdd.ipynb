```python
# Versão do Spark Context
from pyspark.sql import SparkSession

spark = SparkSession.builder \
    .appName("MeuPrimeiroRDD") \
    .getOrCreate()

print(spark.version)
```

```python
#Criando um RDD com 6 elemntos, seria a transformação de ma lista em puthon para um RDD paralelizado 
paises = ["Brasil","Italia","Russia","Noruega","Espanha","Mexico"]
dadosrdd=spark.sparkContext.parallelize(paises)
```

```python
#Exibição de uma coleção, ou seja, todos os dados gravados no RDD sobre paises
dadosrdd.collect()
```

```python
# conta a quantidade de elementos do RDD sobre paises
dadosrdd.count()
```

```python
# exibindo a quantidade de partições
dadosrdd2 = spark.sparkContext.parallelize(paises)
print("Partições: "+str(dadosrdd2.getNumPartitions()))
```

```python
# forçando a criação de 3 partições
dadosrdd3 = spark.sparkContext.parallelize(paises,3)
print("Partições: "+str(dadosrdd3.getNumPartitions()))
```

```python
# carregando um arquivo externo para um RDD
food = spark.sparkContext.textFile('../data/food_coded.csv')
```

```python
# mostrando os 5 primeiros registros do RDD
food.take(5)
```

```python
# Será criado uma RDD lendo dados de um arquivo .csv de uma forma diferente, utilizanso o Sparn Context (sc)
#  Arquivo sobre dados de alimentos
food2 = spark.sparkContext.textFile("../data/food_coded.csv")
```

```python
food2.take(5)
```

```python
#Salvando os dados na pasta, extraindo dados do RDD para um arquivo

GuardaRDD = open("../output/5primfood.csv","w")
GuardaRDD.write("\n".join(food.take(5)))
GuardaRDD.close()
```

```python
# exibindo o primeiro dado dentro do RDD
food.first()
```

```python
# determinando a quantidade de elementos, subtraindo de -1 para retirar o cabeçalho do arquivo
food.count()-1
```

```python
# Pesquisa uma palavra dentro do grupo de dados
filtro_palavra = food.filter(lambda x: 'Arabic cuisine' in x)
filtro_palavra.collect()
```

```python
# persistindo os dados em memória para ficar mais rápido, se houver muitos dados isso acelerará a pesquisa
food.cache()
```

```python
# fechando a sessão do Spark
spark.stop()
```
